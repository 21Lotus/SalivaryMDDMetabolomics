```python
#Creating a new df with depression label and sample
df_new = df[['sample_name', 'poms_depression3']]
df_new = df_new[df_new['poms_depression3'] != 'not applicable']
df_new
```

```python
#Processing the datatable to look nicer
df_new['sample_name'] = df_new['sample_name'].str.replace("11259.", '', regex=False)
df_new['sample_name'] = df_new['sample_name'].str.replace(".", '_', regex=False)
df_new['sample_name'] = df_new['sample_name'].astype(str) + ".mzML"
df_new
```

```python
#normalizing all depression scores

def categorize_depression(value):
  if value=="0":
    return 0
  if value=="1":
    return 1
  if value=="2":
    return 2
  if value=="3":
    return 3
  else:
    return 4

df_new['diagnosis'] = df_new['poms_depression3'].apply(categorize_depression)

print(df_new['diagnosis'].value_counts())
```

```python
df_new = df_new.rename(columns={'diagnosis': "diagnosis", "sample_name": "file_name"})
df_new
```

```python
df_new = df_new.drop(columns=['poms_depression3'])
df_new
```
