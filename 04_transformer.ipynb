```python
#Apply LogFC for best results
import numpy as np
import pandas as pd

def logfc_transform(X, y, control_label=0, eps=1e-6):
  y = pd.Series(y, index=X.index)
  control_ref = X[y == control_label].median(axis=0)
  X_logfc = np.log2((X + eps) / (control_ref + eps))
  return X_logfc, control_ref
```

```python
#TabTransformer with modification
class TabTransformer(nn.Module):
    def __init__(self, num_features, dim_embedding=32, num_heads=1, num_layers=2):
        super(TabTransformer, self).__init__()
        self.embedding = nn.Linear(num_features, dim_embedding)
        encoder_layer = nn.TransformerEncoderLayer(d_model=dim_embedding, nhead=num_heads, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.classifier = nn.Linear(dim_embedding, 1)

    def forward(self, x):
        x = self.embedding(x)
        x = x.unsqueeze(1)  # Adding a sequence length dimension
        x = self.transformer(x)
        x = torch.mean(x, dim=1)  # Pooling
        x = self.classifier(x)
        x = x.squeeze(-1)
        return x
```

```python
#Set seed
torch.manual_seed(42)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Converting data to tensors
X_train_tensor = torch.FloatTensor(X_train_scaled).to(device)
y_train_tensor = torch.FloatTensor(y_train.values).to(device)
X_val_tensor = torch.FloatTensor(X_val_split).to(device)
y_val_tensor = torch.FloatTensor(y_val_split.values).to(device)

criterion = nn.BCEWithLogitsLoss()

model = TabTransformer(248).to(device)
optimizer = optim.Adam(model.parameters(), lr=0.001, betas=(0.9, 0.999), eps=1e-8)

#storing metrics
train_losses = []
train_accs = []
val_losses = []
val_accs = []

#compute probabilities from logits
def compute_accuracy(predictions, targets):
  with torch.no_grad():
    probs = torch.sigmoid(predictions)
    preds = (probs > 0.1).float()
    accuracy = (preds == targets).float().mean()
  return accuracy.item()
```

```python
# Original TabTransformer (working model)
class TabTransformer(nn.Module):
  def __init__(self, num_features, dim_embedding=32, num_heads=1, num_layers=2):
    super(TabTransformer, self).__init__()
    self.embedding = nn.Linear(num_features, dim_embedding)
    encoder_layer = nn.TransformerEncoderLayer(d_model=dim_embedding, nhead=num_heads, batch_first=True)
    self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
    self.classifier = nn.Linear(dim_embedding, 1)

  def forward(self, x):
    x = self.embedding(x)
    x = x.unsqueeze(1)
    x = self.transformer(x)
    x = torch.mean(x, dim=1)
    x = self.classifier(x)
    x = x.squeeze(-1)
    return x

# MultiModal TabTransformer with metabolite embeddings
class MultiModalTabTransformer(nn.Module):
  def __init__(self, num_features, metabolite_embeddings, dim_embedding=32, num_heads=1, num_layers=2):
  super(MultiModalTabTransformer, self).__init__()
  # Store metabolite embeddings
  self.register_buffer('metabolite_emb', metabolite_embeddings)  # Shape: (248, 480)
  metabolite_dim = metabolite_embeddings.shape[1]
  # Project metabolite embeddings to lower dimension
  self.metabolite_projection = nn.Linear(metabolite_dim, dim_embedding)
  # Tabular processing (same as original)
  self.embedding = nn.Linear(num_features, dim_embedding)
  # Transformer
  encoder_layer = nn.TransformerEncoderLayer(d_model=dim_embedding, nhead=num_heads, batch_first=True)
  self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
  # Classifier
  self.classifier = nn.Linear(dim_embedding, 1)

  def forward(self, x):
    batch_size = x.shape[0]  # x shape: [batch_size, 248]
    x_emb = self.embedding(x)  # Process tabular features, [batch, dim_embedding]
    metabolite_emb = self.metabolite_projection(self.metabolite_emb)  # Process metabolite embeddings, [248, dim_embedding]

    # Expand for batch and combine
    x_emb = x_emb.unsqueeze(1)  # [batch, 1, dim_embedding]
    metabolite_emb = metabolite_emb.unsqueeze(0).expand(batch_size, -1, -1)  # [batch, 248, dim_embedding]
    combined = x_emb.expand(-1, 248, -1) + metabolite_emb  # [batch, 248, dim_embedding]
    transformed = self.transformer(combined) #apply transformer
    pooled = torch.mean(transformed, dim=1)  # Pool across metabolites
    output = self.classifier(pooled) # Classify

    return output.squeeze(-1)

# Enhanced MLP with metabolite information
class EnhancedMLP(nn.Module):
  def __init__(self, num_features, metabolite_embeddings):
  super(EnhancedMLP, self).__init__()
  # Use metabolite embeddings to create feature importance weights
  metabolite_dim = metabolite_embeddings.shape[1]
  self.metabolite_encoder = nn.Sequential(nn.Linear(metabolite_dim, 64), nn.ReLU(), nn.Linear(64, num_features), nn.Sigmoid())

  # Register metabolite embeddings
  self.register_buffer('metabolite_emb', metabolite_embeddings)

  # MLP for classification
  self.mlp = nn.Sequential(nn.Linear(num_features, 128),
                           nn.BatchNorm1d(128),
                           nn.ReLU(),
                           nn.Dropout(0.3),
                           nn.Linear(128, 64),
                           nn.BatchNorm1d(64),
                           nn.ReLU(),
                           nn.Dropout(0.3),
                           nn.Linear(64, 32),
                           nn.BatchNorm1d(32),
                           nn.ReLU(),
                           nn.Dropout(0.2),
                           nn.Linear(32, 1)
                          )

  def forward(self, x):
    # Get metabolite-informed weights
    metabolite_features = self.metabolite_encoder(self.metabolite_emb.mean(dim=0, keepdim=True))
    weights = metabolite_features.squeeze(0)

    # Apply weights to input features
    weighted_x = x * weights

    # Classify
    output = self.mlp(weighted_x)
    return output.squeeze(-1)
```

```python
class EnsembleVotingSystem:
  def __init__(self, models, model_names=None, weights=None):
    self.models = models
    self.model_names = model_names or [f"Model_{i}" for i in range(len(models))]
    self.weights = weights or [1.0] * len(models)

  #Get probability predictions from all models
  def predict_proba(self, X_data):
    all_probs = []
    for model in self.models:
      model.eval()
      with torch.no_grad():
        logits = model(X_data)
        probs = torch.sigmoid(logits).cpu().numpy()
        all_probs.append(probs)

    return np.array(all_probs)  # shape: (n_models, n_samples)

  #Make final prediction using voting system
  def predict(self, X_data, threshold=0.5, voting='soft'):
    all_probs = self.predict_proba(X_data)
    if voting == 'soft':
      # Weighted average of probabilities
      weighted_probs = np.average(all_probs, axis=0, weights=self.weights)
      return (weighted_probs > threshold).astype(int)
    elif voting == 'hard':
      # Majority voting
      predictions = (all_probs > threshold).astype(int)
      weighted_votes = np.average(predictions, axis=0, weights=self.weights)
      return (weighted_votes > 0.5).astype(int)
    elif voting == 'stacking':
      # Simple averaging
      return (all_probs.mean(axis=0) > threshold).astype(int)

class ConfidenceWeightedEnsemble(EnsembleVotingSystem):
  def __init__(self, models, model_names=None):
    super().__init__(models, model_names)
    self.calibration_factors = []

  #Calibrate weights based on validation performance
  def calibrate_weights(self, X_val, y_val):
    val_accuracies = []
    for i, model in enumerate(self.models):
      model.eval()
      with torch.no_grad():
        logits = model(X_val)
        probs = torch.sigmoid(logits)
        preds = (probs > 0.5).float()
        acc = (preds == y_val).float().mean()
        val_accuracies.append(acc.item())

    # Use accuracy as weights
    self.weights = np.array(val_accuracies) / sum(val_accuracies)

    print("Model weights based on validation accuracy:")
    for name, weight in zip(self.model_names, self.weights):
      print(f"  {name}: {weight:.3f}")
```

```python
#Train a single model with early stopping
def train_model(model, X_train, y_train, X_val, y_val, criterion, optimizer, n_epochs=100, patience=20):
  best_val_acc = 0
  patience_counter = 0
  best_model_state = None

  train_losses = []
  val_losses = []
  val_accuracies = []

  for epoch in range(n_epochs):
    # Training
    model.train()
    optimizer.zero_grad()

    output = model(X_train)
    loss = criterion(output, y_train)
    loss.backward()
    optimizer.step()

    # Validation
    model.eval()
    with torch.no_grad():
      val_output = model(X_val)
      val_loss = criterion(val_output, y_val)
      val_probs = torch.sigmoid(val_output)
      val_preds = (val_probs > 0.5).float()
      val_acc = (val_preds == y_val).float().mean()

    train_losses.append(loss.item())
    val_losses.append(val_loss.item())
    val_accuracies.append(val_acc.item())

    # Early stopping check
    if val_acc > best_val_acc:
      best_val_acc = val_acc
      patience_counter = 0
      best_model_state = model.state_dict().copy()
    else:
      patience_counter += 1

    if patience_counter >= patience:
      print(f"  Early stopping at epoch {epoch+1}")
      break

    if (epoch + 1) % 20 == 0:
      print(f"  Epoch {epoch+1}: Train Loss: {loss.item():.4f}, "
            f"Val Loss: {val_loss.item():.4f}, Val Acc: {val_acc.item():.4f}")

  # Load best model
  model.load_state_dict(best_model_state)
  return model, train_losses, val_losses, val_accuracies
```
